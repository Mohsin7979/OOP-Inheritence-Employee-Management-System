**🏦 Practice Project: Employee Management System**

Let's create an Employee Management System using inheritance.

Project Requirements

Create:

A parent class Employee,
Child class Developer,
Child class Manager,
Common employee information:
name
salary
A common method:
display_info()
Developer-specific method:
write_code()
Manager-specific method:
manage_team()

**Step 1: Create Parent Class**

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_info(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
```

**Step 2: Create Developer Class**

```python
class Developer(Employee):

    def write_code(self):
        print(self.name, "is writing Python code.")
```

**Step 3: Create Manager Class**

```python
class Manager(Employee):

    def manage_team(self):
        print(self.name, "is managing the team.")
```

**Step 4: Create Objects**

```python
developer = Developer("Ali", 80000)
manager = Manager("Ahmed", 120000)
```

**Step 5: Call Methods**

```python
developer.display_info()
developer.write_code()

print()

manager.display_info()
manager.manage_team()
```

**Step 6: Complete Project Solution**

```python
# Parent Class
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_info(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


# Child Class - Developer
class Developer(Employee):

    def write_code(self):
        print(self.name, "is writing Python code.")


# Child Class - Manager
class Manager(Employee):

    def manage_team(self):
        print(self.name, "is managing the team.")


# Create Developer Object
developer = Developer("Ali", 80000)

# Create Manager Object
manager = Manager("Ahmed", 120000)


# Developer Information
print("===== Developer =====")
developer.display_info()
developer.write_code()


# Manager Information
print("\n===== Manager =====")
manager.display_info()
manager.manage_team()
```
